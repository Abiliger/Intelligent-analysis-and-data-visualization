```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis
from sklearn.datasets import load_wine
```

This code fragment initializes the environment by importing the core Python libraries required for statistical analysis, data processing, and visualization. These imports form the foundation for the analytical part of the project.

```python
wine = load_wine()
df_wine = pd.DataFrame(data=wine.data, columns=wine.feature_names)
df_wine['target'] = wine.target
df_wine.head()
```

This code fragment imports the Wine Dataset from Scikit-learn’s sample data collection and converts it into a structured Pandas DataFrame for analysis.
The dataset (wine) contains chemical composition measurements of different wine samples, along with their corresponding class labels. Using pd.DataFrame(), the numerical data is transformed into a table format with descriptive column names taken from wine.feature_names.
A new column, 'target', is then added to represent the wine class (0, 1, or 2), corresponding to different grape cultivars. Finally, the command df_wine.head() displays the first few rows of the dataset, providing a quick overview of its structure and contents.

```python
climate_data_url = "https://datahub.io/core/co2-ppm/r/co2-mm-mlo.csv"
df_climate = pd.read_csv(climate_data_url)
df_climate.head()
```

This code fragment retrieves and loads an open-source climate dataset containing atmospheric CO₂ concentration measurements.
The variable climate_data_url stores the link to a CSV file hosted on DataHub, which provides monthly CO₂ readings recorded at Mauna Loa Observatory.
Using the pd.read_csv() function from Pandas, the dataset is downloaded directly into a DataFrame named df_climate. This allows for efficient access, inspection, and analysis of time-series climate data.
Finally, the command df_climate.head() displays the first few records, offering a quick preview of the dataset’s structure and contents before further analysis or visualization.

```python
wine_feature = 'alcohol'
climate_feature = 'Interpolated'
```

In this step, two specific variables are defined to identify the features of interest for subsequent analysis.

*   wine_feature = 'alcohol' specifies the alcohol content column from the Wine dataset, which will be analyzed to explore its statistical properties and potential correlations with other features.
*   climate_feature = 'Interpolated' designates the Interpolated CO₂ concentration column from the climate dataset, representing adjusted monthly CO₂ levels.

By assigning these variables, the code ensures a clear and consistent reference to the selected attributes across later analysis and visualization steps.

```python
wine_stats = {
    "Mean": df_wine[wine_feature].mean(),
    "Median": np.median(df_wine[wine_feature]),
    "Variance": np.var(df_wine[wine_feature]),
    "Standard deviation": np.std(df_wine[wine_feature]),
    "Skewness": skew(df_wine[wine_feature]),
    "Kurtosis": kurtosis(df_wine[wine_feature])
}
```

This code fragment creates a dictionary called wine_stats that summarizes the main statistical properties of the selected wine feature — in this case, alcohol content.
Using NumPy and SciPy functions, the code calculates several key indicators that describe how the data is distributed and how much it varies.

*   Mean – shows the average alcohol value across all samples, giving a general sense of the central level.
*   Median – represents the middle value when all alcohol levels are sorted, providing a measure that is less affected by outliers.
*   Variance – measures how spread out the alcohol values are from the mean.
*   Standard deviation – expresses that spread in the same units as the data, making it easier to interpret how much variation exists.
*   Skewness – indicates whether the data leans to the left or right; a measure of how asymmetric the distribution is.
*   Kurtosis – reflects how peaked or flat the distribution is compared to a normal (bell-shaped) curve.

```python
climate_stats = {
    "Mean": df_climate[climate_feature].mean(),
    "Median": np.median(df_climate[climate_feature]),
    "Variance": np.var(df_climate[climate_feature]),
    "Standard deviation": np.std(df_climate[climate_feature]),
    "Skewness": skew(df_climate[climate_feature]),
    "Kurtosis": kurtosis(df_climate[climate_feature])
}
```

This code fragment calculates a set of key statistical measures for the selected climate feature — the Interpolated CO₂ concentration values from the Mauna Loa dataset.
Using the same functions from NumPy and SciPy, it creates a dictionary named climate_stats that summarizes the main characteristics of the data distribution.

*   Mean – shows the average atmospheric CO₂ level during the observed period.
*   Median – indicates the central CO₂ value, giving a balanced view that is less sensitive to extreme monthly variations.
*   Variance – measures how much the CO₂ readings fluctuate around the average.
*   Standard deviation – expresses this variability in the same units as the data, helping to understand how stable or changeable the CO₂ concentration is over time.
*   Skewness – reveals whether the distribution is tilted toward higher or lower CO₂ values.
*   Kurtosis – describes how peaked or flat the distribution is, showing whether most months cluster around the mean or are more spread out.

This code fragment prints the calculated statistical measures for the Wine dataset in a clean, readable format.
The line print("Wine Dataset Statistics") outputs a simple header to indicate which dataset the results refer to.
Then, a for loop iterates through the wine_stats dictionary, where each key represents the name of a statistical metric (such as Mean, Variance, or Skewness), and each value is the corresponding numerical result, formats each number to four decimal places, ensuring the output is precise yet easy to read.

This code fragment generates a histogram to visualize the distribution of the selected climate feature — the Interpolated CO₂ concentration from the Mauna Loa dataset.

A figure of size (10, 5) is created to maintain a wide and clear layout. The sns.histplot() function from Seaborn is used to plot the data, dividing it into 15 bins to show how often different CO₂ concentration levels occur.
The kernel density estimate (kde=True) adds a smooth curve that highlights the overall trend and distribution shape. The plot uses a red color with slight transparency (alpha=0.7), making it visually distinct while maintaining clarity when compared with other graphs.

A legend, title, and axis labels are included to improve readability. The resulting histogram provides an intuitive visual summary of how atmospheric CO₂ concentrations vary over time — an essential step for exploring climate change patterns and long-term trends.

```python
def plot_radial_chart(data, feature_name):
    angles = np.linspace(0, 2 * np.pi, len(data), endpoint=False)
    stats_values = list(data.values())
    stats_values.append(stats_values[0])
    angles = np.concatenate((angles, [angles[0]]))


    plt.figure(figsize=(6, 6))
    ax = plt.subplot(111, polar=True)
    ax.plot(angles, stats_values, 'o-', linewidth=2)
    ax.fill(angles, stats_values, alpha=0.25)
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(data.keys(), fontsize=10)
    plt.title(f'Radial Chart for {feature_name}')
    plt.show()
```

This function, plot_radial_chart(data, feature_name), creates a radial (or spider) chart to visualize multiple statistical indicators at once in a circular layout.

First, the function converts the number of data points into evenly spaced angles around a full circle using np.linspace(). The values from the input dictionary (data.values()) are extracted and the first value is repeated at the end to close the loop, forming a continuous shape.

A new figure is then created with a polar coordinate system, and the data is plotted as a connected line (ax.plot()) with small circular markers. The area inside the shape is softly shaded (ax.fill()), making differences between indicators easy to see.
Each axis represents a statistical measure (like mean, variance, or skewness), and set_xticklabels() labels them clearly around the chart.

Finally, a title showing the analyzed feature’s name (for example, alcohol or CO₂ concentration) is added.
This type of visualization provides an intuitive overview of how each statistical parameter compares to the others — revealing strengths, balance, or variability at a glance.

```python
plot_radial_chart(wine_stats, wine_feature)
```
